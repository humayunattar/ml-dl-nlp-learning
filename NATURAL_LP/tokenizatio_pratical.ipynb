```python
from nltk.tokenize import sent_tokenize
import nltk
try:
    sent_tokenize(corpus)
except LookupError:
    nltk.download('punkt', quiet=True)
    try:
        sent_tokenize(corpus)
    except LookupError:
        nltk.download('punkt_tab', quiet=True)
        sent_tokenize(corpus)
```
