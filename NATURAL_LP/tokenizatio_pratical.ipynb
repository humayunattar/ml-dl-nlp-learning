```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')   # for newer NLTK versions

from nltk.tokenize import word_tokenize

corpus = "I am learning NLP using NLTK."
tokenized_corpus = word_tokenize(corpus)

print(tokenized_corpus)
```

```python
from nltk.tokenize import PunktSentenceTokenizer
corpus = "Hello world! This is a test. Let's see how it works."
sentence_tokenizer = PunktSentenceTokenizer()
sentence_tokenizer.tokenize(corpus)
print(sentence_tokenizer)
```

```python
from nltk.tokenize import TreebankWordTokenizer
corpus = "Hello world! This is a test. Let's see how it works."
word_tokenizer = TreebankWordTokenizer()
tokenized_corpus = word_tokenizer.tokenize(corpus)
print(tokenized_corpus)
```

```python
from nltk.tokenize import sent_tokenize
sent_tokenize(corpus)
print(sent_tokenize(corpus))
```

text preprocessing stemming 

```python
words=["eat","eating","eaten","write","writing","programming","programmer","programmers"]
```

```python
from nltk.stem import PorterStemmer
porter_stemmer = PorterStemmer()
for word in words:
 print(porter_stemmer.stem(word))
```

```python
from nltk.stem import RegexpStemmer
regexp_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
for word in words:
    print(regexp_stemmer.stem(word))
```

```python
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer('english')
for word in words:
    print(snowball_stemmer.stem(word))
```

text preprocessing lemmatization using nltk

```python
from nltk.stem import WordNetLemmatizer
lemmatization = WordNetLemmatizer()
nltk.download('wordnet')


print(lemmatization.lemmatize("programming", pos="v"))
```

```python
for word in words:
    print(lemmatization.lemmatize(word, pos="v"))
```
