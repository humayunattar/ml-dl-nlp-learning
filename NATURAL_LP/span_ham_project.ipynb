```python
import pandas as pd
import nltk
```

```python
messages = pd.read_csv(
    "SMSSpamCollection.txt",
    sep="\t",
    names=["label", "message"]
)
```

```python
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer
from nltk.corpus import stopwords
ps=PorterStemmer()
```

```python
import re
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[ps.stem(word) for word in review if not word in set(stopwords.words('english'))]
    review=' '.join(review)
    corpus.append(review)
```

```python
CV=CountVectorizer(max_features=2500,binary=True,ngram_range=(1,2))
x=CV.fit_transform(corpus).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30,linewidth=1000,formatter=dict(float=lambda x: f'{x:.3f}'))
```

```python
y=pd.get_dummies(messages['label'], drop_first=True)
```

```python
y=y.iloc[:,0].values
```

```python
from sklearn.model_selection import train_test_split
x_train,y_train,x_test,y_test=train_test_split(x,y,test_size=0.20,random_state=0)
```

```python
from sklearn.naive_bayes import MultinomialNB
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=0)
span_detect_model = MultinomialNB().fit(x_train, y_train)
y_pred = span_detect_model.predict(x_test)
y_pred=span_detect_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```
