```python
import pandas as pd
```

```python
#series 
#pandas series is one dimension array like onject that can hold any data type 
data=[1,85,9,6,54]
series=pd.Series(data)
print(series)
```

```python
# create the series from dictionary

data={"a":12,"b":45,"c":78}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[12,5,6,48,56,8]
index=["a","b","c","d","E","F"]
series=pd.Series(data,index=index)
print(series)
```

```python
#dataframe
#create a dataframe from dictionary of list
data={"name":["irfan","humayun","attar"],
      "age":[12,5,2,],"city":["gulbarga","kalburgi","delhi"]}
df=pd.DataFrame(data)
print(df)
```

```python
#create a dadtframe from list of dictionary
data=[{"name":"irfan","age":23,"city":"gul"},{"name":"irfan","age":23,"city":"gul"},{"name":"irfan","age":23,"city":"gul"}]
df=pd.DataFrame(data)
print(df)
```
