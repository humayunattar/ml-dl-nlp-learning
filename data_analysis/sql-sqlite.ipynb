```python
import sqlite3

connection = sqlite3.connect("sales.db")
cursor = connection.cursor()

cursor.execute('''
CREATE TABLE IF NOT EXISTS sale_data (
    date TEXT,
    product TEXT,
    sales INTEGER,
    region TEXT
)
''')
sale_data = [
    ("20-10-23","product1",100,"north"),
    ("22-10-23","product2",100,"south"),
    ("28-10-23","product4",100,"north"),
    ("29-10-23","product5",100,"east"),
    ("24-10-23","product6",100,"west")
]

cursor.executemany('''
INSERT INTO sale_data (date, product, sales, region)
VALUES (?, ?, ?, ?)
''', sale_data)

connection.commit()
```
