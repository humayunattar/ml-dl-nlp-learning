```python
import numpy as np
```

```python
#creating theone dimensional array
arr1=np.array([1,2,3,4,5,])
print(arr1)
print(type(arr1))
print(arr1.shape)
```

```python
# nested array
arr3=np.array([[1,2,3,4,5,],[5,9,7,8,9]])
print(arr3)
```

```python
#array with inbuilt function
np.arange(1,10,2)
```

```python
print(arr3.ndim )# number of dimension 
print(arr3.size ) # number of element
print(arr3.dtype) # return the datatype of the array
print(arr3.itemsize) #return item size in bytes 4,8
```

```python
#numpy vectorized operation
# element vise addition
# element vise subtraction
# element vise multiplication
# element vise division
arr1=np.array([1,2,6,5,9])
arr2=np.array([5,8,9,7,9])
print(arr1+arr2)
print(arr1-arr2)
print(arr1*arr2)
print(arr1/arr2)
```

```python
# universal function 
arr=np.array([1,8,9,7,5])
print(np.sqrt(arr))
```

```python
# exponential
print(np.exp(arr))
```

```python
print(np.sin(arr))
```

```python
print(np.log(arr))
```
