```python
x=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y=df["total_bill"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=10)
```

```python
x_train
```

```python
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
x_train['sex']=le1.fit_transform(x_train['sex'])
x_train['smoker']=le2.fit_transform(x_train['smoker'])
x_train['time']=le3.fit_transform(x_train['time'])
```

```python
x_test['sex']=le1.transform(x_test['sex'])
x_test['smoker']=le2.transform(x_test['smoker'])
x_test['time']=le3.transform(x_test['time'])
```

```python
x_test.head()
```

```python
#onehot encoding
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),['day'])],remainder="passthrough")
```

```python
ct.fit_transform(x_train)
```

```python
import sys
import numpy
numpy.set_printoptions(threshold=sys.maxsize)
x_test=ct.transform(x_test)
```

```python
from sklearn.svm import SVR
svr=SVR()
x_train=ct.fit_transform(x_train)
```

```python
svr.fit(x_train,y_train)
y_pred=svr.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
r2=r2_score(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
print(r2)
print(mae)
```

```python
from sklearn.model_selection import GridSearchCV
grid_param = {
    "C": [0.1, 1, 10, 100, 1000],
    "gamma": [1, 0.1, 0.01, 0.0001],
    "kernel": ["rbf"]
}

grid = GridSearchCV(estimator=SVR(), param_grid=grid_param, refit=True, verbose=2)
grid.fit(x_train, y_train)

print(grid.best_params_)
print(grid.best_score_)
```

```python
print(grid.best_params_)
```

```python
grid_predict=grid.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
r2=r2_score(y_test,grid_predict)
mae=mean_absolute_error(y_test,grid_predict)
print(r2)
print(mae)
```
