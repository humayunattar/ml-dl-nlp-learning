```python
import pandas as pd  
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
from sklearn.datasets import make_classification
x,y=make_classification(n_samples=1000,n_features=2,n_classes=2,n_clusters_per_class=2,n_redundant=0)
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
pd.DataFrame(x)[0]
```

```python
sns.scatterplot(x=pd.DataFrame(x)[0], y=pd.DataFrame(x)[1], hue=y)
```

```python
from sklearn.svm import SVC
svc=SVC(kernel='linear')
```

```python
svc.fit(x_train,y_train)
```

```python
y_pred=svc.predict(x_test)
```

```python
print(y_pred)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
cm=confusion_matrix(y_test,y_pred)
cr=classification_report(y_test,y_pred)
print(cm)
print(cr)
```

```python
rbf=SVC(kernel="rbf")
```

```python
rbf.fit(x_train,y_train)
```

```python
y_pred2=rbf.predict(x_test)
```

```python
cm=confusion_matrix(y_test,y_pred2)
cr=classification_report(y_test,y_pred2)
print(cm)
print(cr)
```

```python
polynomial=SVC(kernel="poly")
polynomial.fit(x_train,y_train)
y_pred3=polynomial.predict(x_test)
```

```python
cm=confusion_matrix(y_test,y_pred3)
cr=classification_report(y_test,y_pred3)
print(cm)
print(cr)
```

```python
sigmoid=SVC(kernel="sigmoid")
sigmoid.fit(x_train,y_train)
y_pred4=sigmoid.predict(x_test)
cm=confusion_matrix(y_test,y_pred4)
cr=classification_report(y_test,y_pred4)
print(cm)
print(cr)
```

```python
# hyperparameter tunning 
from sklearn.model_selection import GridSearchCV
grid_param = {
    "C": [0.1, 1, 10, 100, 1000],
    "gamma": [1, 0.1, 0.01, 0.0001],
    "kernel": ["rbf"]
}

grid = GridSearchCV(estimator=SVC(), param_grid=grid_param, refit=True, verbose=2)
grid.fit(x_train, y_train)

print(grid.best_params_)
print(grid.best_score_)
```
