```python
class Bankaccount:
    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
        print(f"the balance after deposit is {self.balance}")
    def withdrawl(self,amount):
        if amount>self.balance:
            print("insufficient balance")
        else:
            self.balance-=amount
            print(f"the balance after withdrawl is {self.balance}")
    def get_balance(self):
        print(f"the balance is {self.balance}")
account1=Bankaccount("irfan",500000)
account1.deposit(12000)
account1.withdrawl(2000)
account1.get_balance()
```

```python
# single inheritance  
class car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
    def start(self):
        print(f"the persorn start the car with {self.enginetype} engine")
class tesla(car):
    def __init__(self,windows,doors,enginetype,self_driving):
        super().__init__(windows,doors,enginetype)
        self.self_driving=self_driving
    def selfdriving(self):
        print(f"tesla support self driving{self.self_driving} \n and car has {self.windows} windows and {self.doors} doors and {self.enginetype} engine")
tesla1=tesla(4,5,"electric","yes")
tesla1.selfdriving()
tesla1.start()
```

```python
# multiple inheritance
# when a class is derived from more than one base class it is called multiple inheritance
# differnece is we use class name instead of super keyword
class Animal:
    def __init__(self,name):
        self.name=name
class pet:
        def __init__(self,owner):
            self.owner=owner
class dog(Animal,pet):
     def __init__(self, name,owner):
          Animal.__init__(self,name)
          pet.__init__(self,owner=owner)
     def speak(self):
        print(f"{self.name} is a pet of {self.owner} and it says woof")

dog1=dog("tommy", "irfan")
dog1.speak()     
```

```python
#polymorphism 
# metheod overiding example
class shape:
    def area(self):
        return "the AREA OF THE FIGURE"
class rectangle(shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height
    def area(self):# calling the same method name more than once for diifferent action is called polymorphism 
        return self.width*self.height
class circle(shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14*self.radius**2
def print_area(shape):
    print (f"the area is {shape.area()}") # this is the function where all method of classed are called 
rect=rectangle(5,10)
circ=circle(7)
print_area(rect)
print_area(circ)
```

```python
class person:
    def __init__(self,name,age):
        self._name=name
        self._age=age  # protected varaibles
    
class employee(person):
    def __init__(self,name,age,gender):
        super().__init__(name,age)
        self.gender=gender
employee1=employee("irfan", 25, "male")
print(employee1._name)
print(employee1._age)  
print(employee1.gender)
```

```python
#encapsulation with getter and setter methods
class person:
    def __init__(self,name,age):
        self.__name=name
        self.__age=age  # private varaibles
    def get_name(self):
        return self.__name # getter method
    def get_age(self):
        return self.__age # getter method
    def set_name(self,name):
        self.__name=name # setter method
    def set_age(self,age):
        self.__age=age # setter method
person1=person("irfan", 25)
print(person1.get_name())  
print(person1.get_age())
person1.set_name("krishnaik")
person1.set_age(30)
print(person1.get_name())
print(person1.get_age())
```

```python
from abc import ABC,abstractmethod
class vehicle(ABC):
    def drive(self):
        return "the  person start the car"
    @abstractmethod
    def start_engine(self):
        pass
class car(vehicle):
    def start_engine(self):
        return "car engine started "
def operated_vehucle(vehicle):
    print(vehicle.start_engine())
    print(vehicle.drive())
car1=car()
print(operated_vehucle(car1))
```

```python
# OPERATers OVERLOADING for defining the behaviour of the operater
class vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self, other):
        return vector (self.x+other.x,self.y+other.y)
    def __sub__(self, other):
        return vector (self.x-other.x,self.y-other.y)
    def __mul__(self, other):
        return vector (self.x*other.x,self.y*other.y)
    def __truediv__(self, other):
        
        return vector (self.x/other.x,self.y/other.y)
    def __eq__(self, value):
        return self.x==value.x and self.y==value.y
    def __str__(self):
       return f"vector({self.x},{self.y}) "
v1=vector(5,4)
v2=vector(6,8)
print(v1+v2)
print(v1-v2)
print(v1*v2)
print(v1/v2)
print(v1==v2)
```
