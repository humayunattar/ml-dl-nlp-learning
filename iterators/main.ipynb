```python
#decorator
def main_welcome(func):
    #anything which define outside the function will be access
    def sub_welcome(): #inside function
        print(" welcome to the pythone ")
        func
        print("please learn these concept properly")
    return sub_welcome()
```

```python
def cource_indroduction():
    print("this is main function")

cource_indroduction()
```

```python
main_welcome(cource_indroduction())
```

```python
@main_welcome(3)
def cource_indroduction():
    print("this is main function")
```

```python
#decorators with argument
def repeat(n):
    def decorator(fun):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                fun(*args,**kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("hello")            
```

```python
say_hello()
```

```python
# simple example of closures
def outer():
    x = 10
    def inner():
        print(x)   # remembers x
    return inner

f = outer()
f()   # prints 10
```
