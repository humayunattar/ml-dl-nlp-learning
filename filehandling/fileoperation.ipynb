```python
# writing a file with overwritten content
with open ("example.txt","w") as file:
    file.write("This is an example of writing to a file.")
```

```python
# reading a file 
with open ("example.txt","r") as file:
    print(file.read())
```

```python
# writing a file without overwriting content
with open ("example.txt","a") as file:
    file.write("\nThis is an additional line.")
```

```python
# reading a file line by line
with open ("example.txt","r") as file:
    for line in file:
        print(line)
```

```python
# writing multiple lines to a file
list=["fist line\n","second line\n","third line\n"]
with open ("example.txt","a") as file:
    file.writelines(list)
```

```python
# copying the content from source file to destination_file
with open("example.txt","r") as source_file:
    content= source_file.read()
with open("destination.txt","w") as dest_file:
    dest_file.write(content)
```

```python
# counting lines, words, and characters in a file
def count_text(file_path):
    with open(file_path, "r") as file:
        text = file.read()
        
        lines = text.splitlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = len(text)
        
        return line_count, word_count, char_count

file_path = "example.txt"
lines, words, char = count_text(file_path)

print(f"{lines} lines, {words} words, {char} characters in the file.")
```

```python
# writing and then reading a file
with open("test.txt","w+") as file:
    file.write("this is example of writing and reading a file")
    file.seek(0)
    print(file.read())
```
