```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
x=6*np.random.rand(100,1)-3
y=0.5*x**2+x+2+np.random.rand(100,1)
plt.scatter(x,y,color="green")
plt.xlabel("x dataset")
plt.ylabel("y dataset")
plt.title("Polynomial Regression")  
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
regression1=LinearRegression()
regression1.fit(x_train,y_train)
```

```python
from sklearn.metrics import r2_score
y_pred=regression1.predict(x_test)
r2_score(y_test,y_pred)
```

```python
plt.plot(x_train,regression1.predict(x_train),color="green")
plt.scatter(x_train,y_train,color="red")
plt.xlabel("x dataset")
plt.ylabel("y dataset")
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2,include_bias=True)
x_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
```

```python
from sklearn.metrics import r2_score
regression1=LinearRegression()
regression1.fit(x_poly,y_train)
y_pred=regression1.predict(x_test_poly)
r2_score(y_test,y_pred)
```

```python

y_new=regression1.predict(x_new_poly)
plt.plot(x_new,y_new,"r.",label="predicted data")
plt.scatter(x_train.flatten(), y_train.flatten(), c="b", marker=".", label="train data")
plt.plot(x_test,y_test,"g.",label="test data")
plt.legend()
plt.plot(x_test,y_test,"g.",label="test data")
plt.legend()
```

```python
from sklearn.pipeline import Pipeline
```

```python
def polynomial_regression(degree):
    x_new=np.linspace(-3,3,200).reshape(200,1)
    polynomial_features=PolynomialFeatures(degree=degree,include_bias=True)
    linear_regression=LinearRegression()
    pipeline=Pipeline([("polynomial_features",polynomial_features),("linear_regression",linear_regression)])
    pipeline.fit(x_train,y_train)
    y_new=pipeline.predict(x_new)
    plt.plot(x_new,y_new,"r.",label="degree="+str(degree))
    plt.plot(x_train,y_train,"b.",linewidth=3)
    plt.plot(x_test,y_test,"g.",linewidth=3)
    plt.legend(loc="upper left")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.axis([-4,4,0,10])    
```

```python
polynomial_regression(5)
```

```python
polynomial_regression(10)
```
